```python
import numpy as np
import pandas as pd
df2 = pd.read_csv("imdb_top_1000.csv")
print(df2)
```

```python
df2.info()
```

```python
df2["Gross"] = df2["Gross"].str.replace(",", "", regex=False).astype("float")
```

```python
print(df2)
```

```python
df2["Gross"] = df2["Gross"].fillna(0)
```

```python
df2["Meta_score"] = df2["Meta_score"].fillna(0) 
```

```python
df2["Certificate"] = df2["Certificate"].fillna("NON") 
```

```python
df2.info()
```

```python
df2['Runtime'] = df2['Runtime'].str.replace(' min', '', regex=False).astype(int)
```

```python
df2['Primary_Genre'] = df2['Genre'].str.split(',').str[0]
```

```python
print(df2)
```

```python
pd.set_option('display.float_format','{:,.2f}'.format)
```

 ## What are the top 10 highest-grossing movies (Gross) in the dataset?

```python
top_10_movies= (df2[["Series_Title", "Gross"]].sort_values(by="Gross", ascending=False).head(10).reset_index(drop=True))
top_10_movies.index=top_10_movies.index+1
print(top_10_movies)
```

## List the top 5 movies ranked by audience score (IMDB_Rating) and compare them to the top 5 movies ranked by critic score (Meta_score). Are the top 5 movies identical on both lists?

```python
top_5_by_IMDB =(df2[["Series_Title", "IMDB_Rating"]].sort_values(by="IMDB_Rating", ascending=False).head(5).reset_index(drop=True))
top_5_by_IMDB.index=top_5_by_IMDB.index+1
print(top_5_by_IMDB)               
```

```python
top_5_by_Meta =(df2[["Series_Title", "Meta_score"]].sort_values(by="Meta_score", ascending=False).head(5).reset_index(drop=True))
top_5_by_Meta.index=top_5_by_Meta.index+1
print(top_5_by_Meta) 
```

`No, the top 5 movies are not identical on both lists.`

## Which 5 directors have directed the highest number of movies present in this top 1000 list? What is the average Gross revenue for each of those directors?

```python
top_5_directors= (
    df2.groupby('Director').agg(Number_of_Movies=('Director', 'count'),
          Average_Gross=('Gross', 'mean')).sort_values(by='Number_of_Movies', ascending=False).head(5)
)
print(top_5_directors)
```

## Which Primary_Genre generates the highest average Gross revenue? Create a horizontal bar chart ranking the top 8 genres by earnings.

```python
top_8_Primary_Genre = df2.groupby("Primary_Genre")["Gross"].mean().sort_values(ascending=False).head(8)
print(top_8_Primary_Genre)
```

```python
import matplotlib.pyplot as plt
top_8_Primary_Genre.plot(x ='Average_Gross', y='Primary_Genre', kind='barh', figsize=(10,6))
plt.title('Top 8 Primary Genres by Average Gross Revenue')
plt.gca().invert_yaxis()
plt.grid(True)
plt.show()
```

## Is there a strong correlation between how many people voted for a movie (No_of_Votes) and how much money it earned (Gross)? Plot a scatter plot to visualize the relationship.

```python
correlation = df2['No_of_Votes'].corr(df2['Gross'])
print("Correlation between No_of_Votes and Gross:", round(correlation, 4))
```

```python
df2.plot(x ='No_of_Votes', y='Gross', kind='scatter', figsize=(10,6))
plt.title('Relationship Between Number of Votes and Gross Revenue')
plt.grid(True)
plt.show()
```

There is a moderate positive correlation (0.63) between the number of votes and gross revenue. Movies with more votes tend to earn higher gross revenue, although the relationship is not perfectly linear.


## Do longer movies (Runtime) tend to receive higher IMDB_Rating scores, or do audiences prefer shorter films? Group runtime into 30-minute bins (e.g., 90–120 mins, 120–150 mins) and compare average ratings.

```python
bins = range(0, int(df2['Runtime'].max()) + 31, 30)
df2['Runtime_Bin'] = pd.cut(df2['Runtime'], bins=bins)
runtime_ratings = (df2.groupby('Runtime_Bin', observed=False)['IMDB_Rating'].mean().dropna())
print(runtime_ratings)
```

```python
runtime_ratings.plot(x='runtime_ratings', y='Runtime_Bin', kind='barh', figsize=(10,6))
plt.title('Average IMDB Rating by Runtime Group')
plt.grid(True)
plt.show()
```
